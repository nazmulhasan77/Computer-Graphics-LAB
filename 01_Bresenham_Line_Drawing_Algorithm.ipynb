# 1. Bresenham Line Drawing Algorithm

**Computer Graphics LAB — Python Implementation**

## Problem Explanation


Bresenham's algorithm draws a straight line on a pixel grid using mainly integer arithmetic.
It chooses the next pixel that is closest to the ideal mathematical line. This makes it fast
and suitable for raster displays.


    **বাংলায় সহজভাবে:**  
    এটি line-এর জন্য সবচেয়ে কাছের pixel বেছে নেয়। Floating-point calculation খুব কম লাগে, তাই raster graphics-এ efficient.

## Algorithm Steps

1. Start from the first endpoint (x1, y1).
2. Calculate the horizontal and vertical distances.
3. Use an error/decision value to decide when x or y must change.
4. Store every selected pixel until the second endpoint is reached.
5. Plot the selected pixels.

## Python Solution

নিচের code cell run করলে algorithm-এর result graphically দেখা যাবে।
Sample input দেওয়া আছে; চাইলে নিজের value বসিয়ে পরিবর্তন করতে পারবেন।

```python
import matplotlib.pyplot as plt

# Sample endpoints
x1, y1 = 2, 3
x2, y2 = 12, 9

def bresenham(x1, y1, x2, y2):
    points = []

    dx = abs(x2 - x1)
    dy = abs(y2 - y1)

    sx = 1 if x1 < x2 else -1
    sy = 1 if y1 < y2 else -1

    err = dx - dy

    while True:
        points.append((x1, y1))

        if x1 == x2 and y1 == y2:
            break

        e2 = 2 * err

        if e2 > -dy:
            err -= dy
            x1 += sx

        if e2 < dx:
            err += dx
            y1 += sy

    return points

points = bresenham(x1, y1, x2, y2)

x = [p[0] for p in points]
y = [p[1] for p in points]

print("Generated pixels:", points)

plt.plot(x, y, marker="s")
plt.title("Bresenham Line")
plt.xlabel("X")
plt.ylabel("Y")
plt.grid(True)
plt.axis("equal")
plt.show()
```
