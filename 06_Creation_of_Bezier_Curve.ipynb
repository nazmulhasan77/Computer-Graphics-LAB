# 6. Creation of Bezier Curve

**Computer Graphics LAB — Python Implementation**

## Problem Explanation


A Bezier curve is a smooth parametric curve controlled by a set of control points.
For a cubic Bezier curve, four control points are used. The curve starts at the first point,
ends at the last point, and is pulled toward the middle control points.


    **বাংলায় সহজভাবে:**  
    Bezier curve control point দিয়ে shape control করা হয়। Curve সাধারণত মাঝের control point-এর উপর দিয়ে যায় না, কিন্তু তাদের দিকে bend করে।

## Algorithm Steps

1. Choose four control points P0, P1, P2, P3.
2. Let parameter t vary from 0 to 1.
3. Use the cubic Bezier formula to calculate each point on the curve.
4. Store the calculated x and y coordinates.
5. Plot both the control polygon and the Bezier curve.

## Python Solution

নিচের code cell run করলে algorithm-এর result graphically দেখা যাবে।
Sample input দেওয়া আছে; চাইলে নিজের value বসিয়ে পরিবর্তন করতে পারবেন।

```python
import matplotlib.pyplot as plt
import numpy as np

# Four control points
P0 = np.array([0.0, 0.0])
P1 = np.array([2.0, 5.0])
P2 = np.array([6.0, 5.0])
P3 = np.array([8.0, 0.0])

def bezier_point(t, P0, P1, P2, P3):
    return (
        (1 - t)**3 * P0
        + 3 * (1 - t)**2 * t * P1
        + 3 * (1 - t) * t**2 * P2
        + t**3 * P3
    )

t_values = np.linspace(0, 1, 200)
curve = np.array([bezier_point(t, P0, P1, P2, P3) for t in t_values])

control = np.array([P0, P1, P2, P3])

plt.plot(control[:, 0], control[:, 1], marker="o", linestyle="--", label="Control points")
plt.plot(curve[:, 0], curve[:, 1], linewidth=2, label="Bezier curve")

plt.title("Cubic Bezier Curve")
plt.xlabel("X")
plt.ylabel("Y")
plt.grid(True)
plt.axis("equal")
plt.legend()
plt.show()
```
