# 4. Sutherland–Hodgman Polygon Clipping Algorithm

**Computer Graphics LAB — Python Implementation**

## Problem Explanation


Sutherland–Hodgman clips a polygon against a convex clipping boundary.
For a rectangular window, the polygon is processed one boundary at a time:
left, right, bottom, and top. New intersection points are inserted when edges cross a boundary.


    **বাংলায় সহজভাবে:**  
    Polygon-এর যে অংশ clipping window-এর বাইরে থাকে তা বাদ যায়। প্রতিটি boundary ধরে একবার করে polygon process করা হয়।

## Algorithm Steps

1. Take the polygon vertices as the input list.
2. Clip the polygon against the left boundary.
3. Use the output as input for the right boundary.
4. Then clip against bottom and top boundaries.
5. Plot the final list of vertices.

## Python Solution

নিচের code cell run করলে algorithm-এর result graphically দেখা যাবে।
Sample input দেওয়া আছে; চাইলে নিজের value বসিয়ে পরিবর্তন করতে পারবেন।

```python
import matplotlib.pyplot as plt

# Rectangular clipping window
xmin, ymin = 2, 2
xmax, ymax = 8, 7

# Sample polygon
polygon = [(1, 3), (4, 1), (9, 3), (7, 8), (3, 7)]

def inside(p, edge):
    x, y = p

    if edge == "left":
        return x >= xmin
    if edge == "right":
        return x <= xmax
    if edge == "bottom":
        return y >= ymin
    if edge == "top":
        return y <= ymax

def intersection(p1, p2, edge):
    x1, y1 = p1
    x2, y2 = p2

    if edge == "left":
        x = xmin
        y = y1 + (y2 - y1) * (xmin - x1) / (x2 - x1)

    elif edge == "right":
        x = xmax
        y = y1 + (y2 - y1) * (xmax - x1) / (x2 - x1)

    elif edge == "bottom":
        y = ymin
        x = x1 + (x2 - x1) * (ymin - y1) / (y2 - y1)

    else:  # top
        y = ymax
        x = x1 + (x2 - x1) * (ymax - y1) / (y2 - y1)

    return (x, y)

def clip_against_edge(points, edge):
    if not points:
        return []

    output = []
    previous = points[-1]

    for current in points:
        prev_inside = inside(previous, edge)
        curr_inside = inside(current, edge)

        if curr_inside:
            if not prev_inside:
                output.append(intersection(previous, current, edge))
            output.append(current)

        elif prev_inside:
            output.append(intersection(previous, current, edge))

        previous = current

    return output

def sutherland_hodgman(polygon):
    result = polygon

    for edge in ["left", "right", "bottom", "top"]:
        result = clip_against_edge(result, edge)

    return result

clipped = sutherland_hodgman(polygon)
print("Clipped polygon:", clipped)

def draw_polygon(points, label, style="-"):
    if not points:
        return
    closed = points + [points[0]]
    xs = [p[0] for p in closed]
    ys = [p[1] for p in closed]
    plt.plot(xs, ys, style, marker="o", label=label)

# Draw clipping rectangle
rect = [(xmin, ymin), (xmax, ymin), (xmax, ymax), (xmin, ymax)]
draw_polygon(rect, "Clipping window")

# Draw original and clipped polygons
draw_polygon(polygon, "Original polygon", "--")
draw_polygon(clipped, "Clipped polygon", "-")

plt.title("Sutherland-Hodgman Polygon Clipping")
plt.xlabel("X")
plt.ylabel("Y")
plt.grid(True)
plt.axis("equal")
plt.legend()
plt.show()
```
