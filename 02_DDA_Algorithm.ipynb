# 2. DDA (Digital Differential Analyzer) Algorithm

**Computer Graphics LAB — Python Implementation**

## Problem Explanation


DDA is a line drawing algorithm. It calculates small equal steps from one endpoint to another.
At each step it adds a fixed amount to x and y, then rounds the result to the nearest pixel.


    **বাংলায় সহজভাবে:**  
    DDA line-কে অনেক ছোট step-এ ভাগ করে। প্রতি step-এ x ও y increment যোগ করে নতুন pixel বের করে।

## Algorithm Steps

1. Calculate dx = x2 - x1 and dy = y2 - y1.
2. Set steps = max(|dx|, |dy|).
3. Calculate x_increment = dx / steps and y_increment = dy / steps.
4. Start from (x1, y1) and repeatedly add the increments.
5. Round the coordinates and plot the generated pixels.

## Python Solution

নিচের code cell run করলে algorithm-এর result graphically দেখা যাবে।
Sample input দেওয়া আছে; চাইলে নিজের value বসিয়ে পরিবর্তন করতে পারবেন।

```python
import matplotlib.pyplot as plt

# Sample endpoints
x1, y1 = 2, 3
x2, y2 = 12, 9

def dda(x1, y1, x2, y2):
    dx = x2 - x1
    dy = y2 - y1

    steps = max(abs(dx), abs(dy))

    if steps == 0:
        return [(round(x1), round(y1))]

    x_inc = dx / steps
    y_inc = dy / steps

    x = x1
    y = y1
    points = []

    for _ in range(steps + 1):
        points.append((round(x), round(y)))
        x += x_inc
        y += y_inc

    return points

points = dda(x1, y1, x2, y2)

x = [p[0] for p in points]
y = [p[1] for p in points]

print("Generated pixels:", points)

plt.plot(x, y, marker="o")
plt.title("DDA Line")
plt.xlabel("X")
plt.ylabel("Y")
plt.grid(True)
plt.axis("equal")
plt.show()
```
