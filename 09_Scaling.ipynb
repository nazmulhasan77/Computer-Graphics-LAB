# 9. 2D Scaling

**Computer Graphics LAB — Python Implementation**

## Problem Explanation


Scaling changes the size of an object.
If sx and sy are the horizontal and vertical scaling factors, then:
x' = sx * x
y' = sy * y
A factor greater than 1 enlarges the object, while a factor between 0 and 1 shrinks it.


    **বাংলায় সহজভাবে:**  
    Scaling object-এর size বড় বা ছোট করে। sx এবং sy দিয়ে x ও y direction-এ কত গুণ change হবে তা নির্ধারণ করা হয়।

## Algorithm Steps

1. Take the object coordinates.
2. Choose scaling factors sx and sy.
3. Multiply each x-coordinate by sx.
4. Multiply each y-coordinate by sy.
5. Plot the original and scaled object.

## Python Solution

নিচের code cell run করলে algorithm-এর result graphically দেখা যাবে।
Sample input দেওয়া আছে; চাইলে নিজের value বসিয়ে পরিবর্তন করতে পারবেন।

```python
import matplotlib.pyplot as plt
import numpy as np

shape = np.array([
    [1, 1],
    [4, 1],
    [2.5, 4],
    [1, 1]
], dtype=float)

sx, sy = 1.5, 2.0

scaled = shape.copy()
scaled[:, 0] *= sx
scaled[:, 1] *= sy

print("Original points:\n", shape)
print("Scaled points:\n", scaled)

plt.plot(shape[:, 0], shape[:, 1], marker="o", label="Original")
plt.plot(scaled[:, 0], scaled[:, 1], marker="o", label="Scaled")

plt.title("2D Scaling About Origin")
plt.xlabel("X")
plt.ylabel("Y")
plt.grid(True)
plt.axis("equal")
plt.legend()
plt.show()
```
